```python
food_list = ['Bananas', 'Chocolate', 'bread', 'diapers', 'Ice Cream', 'Brownie Mix', 'broccoli']

#Create a new list with the words that begin w/b
b_words = []

for item in food_list:
    if item.startswith('b'):
        b_words.append(item)

print(b_words)
print(type(b_words))

#Create the final grocery list, including a comma and a space between each item, w/period at the end & ensuring all are lower case.
grocery_list = ("Grocery list: " + ", ".join(b_words).lower() + ".")
print(grocery_list)
```

# Challenge 3 - Regular Expressions

Sometimes, we would like to perform more complex manipulations of our string. This is where regular expressions come in handy. In the cell below, return all characters that are upper case from the string specified below.

```python
poem = """The apparition of these faces in the crowd;
Petals on a wet, black bough."""


def look_for_upper_case(text):
    uppercase_characters = []
    for characters in text:
        if characters.isupper():
            uppercase_characters.append(characters)
    return uppercase_characters
```

```python
look_for_upper_case(poem)
```

In the cell below, filter the list provided and return all elements of the list containing a number. To filter the list, use the `re.search` function. Check if the function does not return `None`. You can read more about the `re.search` function [here](https://docs.python.org/3/library/re.html).

```python
data = ['123abc', 'abc123', 'JohnSmith1', 'ABBY4', 'JANE']

def look_for_number(text):
    list_with_numbers = []
    for item in text:
        if re.search(r'\d', item):
            list_with_numbers.append(item)
    return list_with_numbers
```

```python
look_for_number(data)
```
