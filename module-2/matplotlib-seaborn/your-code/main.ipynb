#### Use your previous code but now, resize your plot.
**Hint**: Add the figsize argument in plt.subplots().

If you want, try to add a title to the plot or even axes labels. You can also play with the fontweight and fontsize of the titles and labels. 

```python
fig, axs = plt.subplots(ncols = 2,
                       nrows = 1,
                       figsize = (9, 5))
axs[0].plot(x, y)                    
axs[1].plot(x, z)

axs[0].set(xlabel='x',
           ylabel='y',
           title='x vs y')
axs[1].set(xlabel='x',
           ylabel='z',
           title='x vs z')

plt.subplots_adjust(wspace=0.4)

plt.show()
```

#### Now you will have to plot both $y=x^2$ and $y=e^x$ in the same plot. Create two subplots and use a normal scale in the first one and a logarithmic scale in the second one.
**Hint**: Use `set_xscale` and `set_yscale`.

```python
y1 = x*2
y2 = np.exp(x)


fig, axs = plt.subplots(ncols = 2,
                       nrows = 1,
                       figsize = (9, 5))
axs[0].plot(y1, y2)                    
axs[1].plot(y1, y2)

axs[0].set(xlabel='y1',
           ylabel='y2',
           title='y1 vs y2')

axs[1].set(xlabel='y1',
           ylabel='y2',
           title='y1 vs y2 log scale')

axs[1].set_xscale("log")
axs[1].set_yscale("log")

plt.subplots_adjust(wspace=0.4)

plt.show()
```

#### Use subplots and plot the distribution of the `Age`  with bins equal to 10, 20 and 50.

```python
# your code here

fig, axes = plt.subplots(nrows=1, ncols=3, figsize=(14, 6))

axes[0].hist(titanic['Age'], bins=10)
axes[0].set_title('Histogram 10 bins')

axes[1].hist(titanic['Age'], bins=20)
axes[1].set_title('Histogram 20 bins')

axes[2].hist(titanic['Age'], bins=50)
axes[2].set_title('Histogram 50 bins')

for ax in axes:
    ax.set_xlabel('Age')
    ax.set_ylabel('Frequency')


plt.show()
```
